```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
test_file_v1 = pd.read_csv("generation_test_v1.csv")
v1_data = pd.DataFrame(test_file_v1)
unreachable_v1 = v1_data["UnreachableRooms"]
unreachable_v1.value_counts()
```

```python
unreachable_v1.value_counts()[1:5].sort_index().plot(kind="bar")
```
